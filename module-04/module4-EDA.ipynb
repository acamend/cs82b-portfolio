#### Exercise 2.1: Load the dataset into a Pandas DataFrame using pd.read_csv()

Definitions:

• RM: The average number of rooms per dwelling

• LSTAT: The percentage of lower-status population in the area

• PTRATIO: The pupil-teacher ratio by town

• MEDV: The median value of owner-occupied homes

```python
df.info()
```

```python
# Inspect using describe()
df.describe()
```

```python
#Change units of MEDV to 1000s and rename MEDV_T
df['MEDV_T'] = df['MEDV']/1000
df.drop(['MEDV'], axis = 1, inplace = True)
df.head()
```

#### Exervise 2.2: Create a scatter plot of RM on the x-axis and MEDV on y-axis
RM: Average number of rooms per dwelling

MEDV: Median value of owner-occupied homes in $1000\'s

Title: Relationship between RM and MEDV

```python
plt.scatter(df['RM'], df['MEDV_T'])
plt.xlabel('Average number of rooms per dwelling')
plt.ylabel('Median value of owner-occupied homes in $1000s')
plt.title('Relationship between RM and MEDV')
```

#### Exervise 2.2: Create visualization using histogram of Median Value

Histogram of MEDV 

X Label: Median value of owner-occupied homes in $1000\'s (MEDV)

Y Label: Frequency 

Title: Histogram of MEDV

```python
sns.histplot(data=df, x="MEDV_T", bins=20)
plt.title("Histogram of MEDV")
plt.xlabel("Median value of owner-occupied homes in $1000's (MEDV)")
plt.ylabel("Frequency")
```

**Comment**
The data is right skewed, as is common for housing prices and incomes.

#### Exercise 2.3 Analyzing the relationship between each independent variable and the dependent variable using .corr() and `print(corr_matrix['MEDV'].sort_values(ascending=False))`



```python
# Use .corr() method

sns.heatmap(df.corr(), annot=True)
```

```python
#Use corr_matrix to show only the MEDV correlations
corr_matrix = df.corr()
print(corr_matrix['MEDV_T'].sort_values(ascending=False))
```

**Results**
Number of rooms has a positive correlation with MEDV whereas pupil/teacher ratio and percent lower status population have a negative correlation with MEDV.

#### Exercise 2.4 Identifying any outliers of MEDV in the dataset using boxplot(). Add a title 'Boxplot of MEDV'


```python
sns.boxplot(y=df['MEDV_T'])
plt.title("Boxplot of MEDV")
```

```python
sns.boxenplot(y=df['MEDV_T'])
plt.title("Boxenplot of MEDV")
```

**Comment**
The box plot shows that the data is right skewed, quite a few high outlier values. The boxen plot shows that there are also some very low outliers.

```python
# Visualization 1 - Check for Daily usage outliers
sns.boxplot(y=df['Daily_Usage_Hours'])
plt.title("Boxplot of Daily Usage Hours")
```

```python
# Visualization 2 - View Histogram for Daily usage outliers
# Provides more detail on box plot outliers
sns.histplot(data=df, x='Daily_Usage_Hours', bins=10)
plt.title("Histogram of Daily Usage Hours")
plt.xlabel("Daily usage hours")
plt.ylabel("Frequency")
```

# Comments on Visualizations 1 and 2
The boxplot shows that the median social media daily usage is about 4.5 hours. There are numerous values above the third quartile, 75th percentile. 

The histogram shows a clear right skew to the data, with large quanitities of responses in each bin. This indicates skew to me, not erroneous values.

```python
#Visualization 3 - Correlation matrix 
# I am interested in how daily usage impacts perceived stress score and sleep duration.

#sns.heatmap(df.corr(), annot=True)
#This did work, must create subset of data with only numeric columns

##   AI Prompt: How do I pull out only numeric columns from a dataset?

# Only calculate correlation for numeric columns
numerical_df = df.select_dtypes(include=['number'])
sns.heatmap(numerical_df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
#Visualization #4 Scatterplot showing - 
#Relationship between Daily Usage Hours and Perceived Stress Score
plt.scatter(df['Daily_Usage_Hours'], df['Perceived_Stress_Score'])
plt.xlabel('Daily Usage (hours)')
plt.ylabel('Perceived Stress Score')
plt.title('As daily usage increases, the perceived stress score increases')
```

```python
#Visualization #5 Scatterplot showing - 
#Relationship between Daily Usage and Sleep Duration
plt.scatter(df['Daily_Usage_Hours'], df['Sleep_Duration_Hours'])
plt.xlabel('Daily Usage (hours)')
plt.ylabel('Sleep Duration (hours)')
plt.title('As daily usage increases, sleep duration decreases')
```
